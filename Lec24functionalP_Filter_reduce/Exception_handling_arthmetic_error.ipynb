# Exception Handling

## 1.Arthmetic Errors

### 1]Zero Division Error

#### program without Exception Handling

```python
num1 = int(input("enter first number"))
num2 = int(input("enter second number"))
result = num1/num2
```

#### program with Exception Handling

```python
num1 = int(input("enter first number"))
num2 = int(input("enter second number"))
try:
  result = num1/num2
  print(result)
except ZeroDivisionError as err:
  print(err)  
```
