# Reduce Function

## import reduce function

```python
from functools import reduce
```

### Q1.add numbers using reduce function

```python
numbers = [1,2,3,4,5]
def add(x,y):
  return x + y
result = reduce(add,numbers)
print(result)  
```

### using lambda

```python
numbers = [1,2,3,4,5]
add = reduce(lambda x , y :x + y,numbers)
print(add)
```
