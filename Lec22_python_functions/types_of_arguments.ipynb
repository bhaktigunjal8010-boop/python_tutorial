# Types of Arguments

## 1.Positional Argument

```python
def greet(name,age):
  print("my name is :",name,"and my age is :",age)
greet("Bhakti",21)
```

```python
def greet(name,age):
  print("my name is :",name,"and my age is :",age)
greet(21,"Bhakti")
```

## 2.Keyword Arguments

```python
def greet(name,age):
  print("my name is :",name,"and my age is :",age)
greet(name ="Bhakti", age =21)
```

```python
def greet(name,age):
  print("my name is :",name,"and my age is :",age)
greet(age = 21,name="Bhakti")
```

## 3.Default Argument

#### 1]without default Argument

```python
def greet(name,age):
  print("my name is",name,"and my age is",age)
greet("Bhakti")
```

### 2]With Default Argument

```python
def greet(name="priya",age=21):
  print("my name is",name,"and my age is",age)
greet("Bhakti")
```

```python
def greet(name="priya",age=21):
  print("my name is",name,"and my age is",age)
greet("Bhakti",25)
```

```python
def greet(name="priya",age=21):
  print("my name is",name,"and my age is",age)
greet(30)
```

```python
def greet(name="Bhakti",age=21):
  print("my name is",name,"and my age is",age)
greet()
```

## 4.Variable length Positinal Argument

```python
def calculate_sum(*args):
  print(args)
  print(type(args))
  return sum(args)
result = calculate_sum(1,2,3)
print(result)
result = calculate_sum(1,2,3,4,5,6)
```

## 5.Variable length keyword Argument

```python
def print_details(**kwargs):
  print(kwargs)
  print(type(kwargs))
  for k,v in kwargs.items():
     print(k,v)

print_details(name ="rahul", age = 21,city = "pune")     
```

```python
def print_details(**kwargs):
  print(kwargs)
  print(type(kwargs))
  for k,v in kwargs.items():
     print(k,v)

print_details(age = 21,name ="rahul", city = "pune") 
```
