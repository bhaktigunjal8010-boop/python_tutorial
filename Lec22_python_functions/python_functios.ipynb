# Python Functions

## 1. using return function

```python
def calculate_sum(a, b):
    sum = a + b
    return sum
result = calculate_sum(10, 20)
```

```python
print(result)
```

## 2.without using return function

```python
def calculate_sum(a, b):
    sum = a + b
    
result = calculate_sum(10, 20)
```

```python
print(result)
```
