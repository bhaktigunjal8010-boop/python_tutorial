# Dictionary Methods

## 1.update

```python
dict1 = {"name": "Alice", "age": 30}
```

```python
print(dict1)
```

```python
dict1.update({"city":"pune", "country":"India"})  # Adding new key-value pairs to dict1
```

```python
print(dict1)
```

```python
dict1["roll_number"] = 10  # Adding a new key-value pair using assignment
```

```python
print(dict1)
```

```python
dict1["city"] = "Mumbai"  # Updating the value of an existing key
```

```python
print(dict1)
```

## 2.pop(key)

```python
dict1 = {"name":"Alice", "age":30, "city":"New York"}
```

```python
print(dict1)
```

```python
dict1.pop("age")  # Removing the key "city" from dict1
```

```python
print(dict1)
```

```python
dict1.pop("city")  # Removing the key "city" from dict1
```

```python
print(dict1)
```

## 3.popitem()

```python
dict1 = {"name":"Alice", "age":30,"roll_no": 12, "city":"New York"}
```

```python
dict1.popitem()
```

```python
print(dict1)
```

## 4.clear()

```python
dict1 = {"name":"Alice", "age":30, "city":"New York"}
```

```python
print(dict1)
```

```python
dict1.clear()  # Clearing all items from dict1
```

```python
print(dict1)
```
