# Basic Dictionary Operations

## 1.Mutability

```python
dict1 = {"laptop":"Dell", "processor":"i5"}
```

```python
print(dict1)
```

```python
dict1["RAM"] = "8GB"  # Adding a new key-value pair to the dictionary
```

```python
print(dict1)
```

## 2.Conversion

### 1]converting tuple into dictionary

```python
tuple1 = (('name', 'John'), ('age', 30), ('city', 'New York'))  #nested tuple
```

```python
dict1 = dict(tuple1)  # Converting the tuple to a dictionary
```

```python
print(dict1)
```

### 2]coverting list into dictionary

```python
list1 =[('name', 'Alice'), ('age', 25), ('city', 'Los Angeles')]  #nested list
```

```python
dict1 = dict(tuple1)  # Converting the tuple to a dictionary
```

```python
print(dict1)
```

### 3]converting set into dictionary

```python
set1 = {('name', 'Bob'), ('age', 35), ('city', 'Chicago')}  #nested set
```

```python
dict1 = dict(list1)  # Converting the list to a dictionary
```

```python
print(dict1)
```

### 4]converting set of elements as key & default values as dictionary

```python
set1 = {"num1","num2","num3"}
```

```python
dict1 = {key :100 for key in set1}  # Creating a dictionary from the set with a default value of 100
```

```python
print(dict1)
```

### 5]converting range as keys & values as default values as default dictionary

```python
rang1 = range(1,6,1)  # Creating a range object from 1 to 5
```

```python
dict1 = {key :10 for key in rang1}  # Creating a dictionary from the range with a default value of 10
```

```python
print(dict1)
```

## 3.Aliasing

```python
dict1 = {"mobile":"Samsung", "model":"Galaxy S21"}
```

```python
dict2 = dict1
```

```python
print(dict2)
```

```python
print(id(dict1))
print(id(dict2))  # Printing the memory address of dict2 to show that it points to the same object as dict1
```

## 4.Cloning

```python
dict1 = {"mobile":"Samsung", "model":"Galaxy S21"}
```

```python
dict2 = dict1.copy()  # Creating a copy of the dictionary
```

```python
print(dict2)
```

```python
print(id(dict1))
print(id(dict2))  # Printing the memory address of dict2 to show that it points to a different object than dict1
```

```python
dict1["price"] = "$799"  # Adding a new key-value pair to dict1
```

```python
print(dict1)
print(dict2)
```

## 5.Searching

```python
dict1 = {"name":"Alice", "age":30, "city":"New York"}
```

```python
print("name" in dict1)  # Checking if the key "name" exists in dict1
```

```python
print("roll_no" in dict1)  # Checking if the key "roll_no" exists in dict1
```

## 6.Identity

```python
dict1 = {"name":"Alice", "age":30}
```

```python
dict2 = dict1
```

```python
print(dict1 is dict2)  # Checking if dict1 and dict2 point to the same object
```

```python
dict3 = { "name":"Alice", "age":30}
```

```python
dict4 = dict3.copy()  # Creating a copy of dict3
```

```python
print(dict3 is dict4)  # Checking if dict3 and dict4 point to the same object as cloning has diff memory address 
```
