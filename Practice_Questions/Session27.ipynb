### 7.Create a class named Rectangle that takes length and breadth as parameters. Add two methods: area() that returns length × breadth and perimeter() that returns 2 × (length + breadth). Create two rectangle objects and print their area and perimeter.

```python

class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length + self.breadth)


object1 = Rectangle(10, 5)
object2 = Rectangle(8, 4)

print("Area of Rectangle 1:", object1.area())
print("Perimeter of Rectangle 1:", object1.perimeter())

print("--------------------")

print("Area of Rectangle 2:", object2.area())
print("Perimeter of Rectangle 2:", object2.perimeter())
```

### 9.Create a class named Pet with attributes name and sound. Add a method speak() that prints "<name> says <sound>". Create objects for a dog, a cat and a cow, and call speak() for each.


```python
class Pet:
  def __init__(self,name,sound):
    self.name = name
    self.sound = sound
  def speak(self):
    print(self.name,self.sound)
    
dog = Pet("dog","bhoo bhoo")
dog.speak() 
cat = Pet("cat","meow meow")
cat.speak()
cow = Pet("cow says","hmm hmmm")  
cow.speak()  
```
