## Variable Length Argument

### Q1.Write a function show_values(*args) that prints args and its type. Call it with the values 1, 2, 3.

```python
def show_values(*args):
  print(args)
  print(type(args))

show_values(1,2,3)  
```

### Q2.Write a function calculate_sum(*args) that returns the sum of all the numbers passed to it. Call it once with 1, 2, 3 and once with 1, 2, 3, 4, 5, 6, 7, and print both results

```python
def calculate_sum(*args):
  print(args)
  return sum(args)
  
calculate_sum(1,2,3)
calculate_sum(1,2,3,4,5,6,7)  
```

### Q3.Write a function count_values(*args) that returns how many values were passed to it. Call it with 5 values and then with no values.

```python
def count_values(*args):
  print(len(args))
  return sum(args)

count_values(1,2,3,4,5)
count_values()  
```

### Q4.Write a function find_max(*args) that returns the largest number passed to it using max(). Call it with 10, 45, 23, 67, 12.

```python
def find_max(*args):
  print(max(args))


find_max(10,45,23,67,12)  
```

### Q5.Write a function print_names(*args) that prints each name on a separate line using a for loop. Call it with "rahul", "amit", "sneha".

```python
def print_names(*args):
  for name in args :
      print(name)

print_names("rahul","amit","sneha")  
```

### Q6.Write a function calculate_average(*args) that returns the average of all the numbers passed to it. Call it with 10, 20, 30, 40.

```python
def calculate_average(*args):
  
  return sum(args) / len(args) 

calculate_average(10,20,30,40)  
```

### Q7.Write a function print_details(**kwargs) that prints kwargs and its type. Call it with name="rahul", age=20, city="pune".

```python
def print_details(**kwargs):
  print(kwargs)
  print(type(kwargs))

print_details(name = "rahul", age = 20 , city = "pune")  
```

### Q8.Using the print_details(**kwargs) function from Q7, add a for loop with items() to print each key and value on a separate line.

```python
def print_details(**kwargs):
  
  for k,v in kwargs.items():
    print( k , ":", v)

print_details(name = "rahul", age = 20 , city = "pune")  
```

### Q9. Write a function student_info(**kwargs) that prints only the keys passed to it using keys(). Call it with name="sneha", roll_no=51, course="python".
