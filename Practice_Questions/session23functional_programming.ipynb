### Q1. Write a normal function summation(x, y) that returns the sum of two numbers. Then write the same logic using a lambda function and print the result for 3 and 5.


```python
##Normal function
def summation(x,y):
  sum = x + y
  return sum

result = summation(3,5)
print(result)  
```

```python
## lambda function
sum = lambda x , y : x + y
print(sum(3,5))
```

### Q2.Write a lambda function that checks whether a number is even. Print the result for 6 and 7.


```python
is_even = lambda x : x % 2 == 0
print(is_even(6))
print(is_even(7))
```

### Q3. Write a lambda function that returns the square of a number. Print the result for 5.


```python
square = lambda x : x ** 2
print(square(5))
```

### Q4.Write a lambda function that returns the cube of a number. Print the result for 3.

### Q5. Write a lambda function that concatenates two strings. Print the result for "hello " and "world".


```python
concat = lambda str1 , str2 : str1 + str2
print(concat("hello ","world!"))
```

### Q6. Write a lambda function that takes three numbers and returns the largest number using max(). Print the result for 10, 45 and 23.


```python
largest = lambda x , y,z : max(x,y,z)
print(largest(10,45,23))
```

### Q7.Create a list numbers = [1, 2, 3, 4, 5]. Write a normal function square(x) and use map() to get the squares of all the numbers. First print the map object directly, then print it after converting it into a list.


```python
numbers = [1,2,3,4,5]
def square (x):
  return x ** 2
sq_nos = map(square,numbers)
print(sq_nos) 
```

```python
numbers = [1,2,3,4,5]
def square (x):
  return x ** 2
sq_nos = map(square,numbers)
print(list(sq_nos)) 
```

### Q8.Using the numbers list from Q7, use map() with a normal function add_ten(x) to add 10 to every value. Then solve the same problem using map() with a lambda function.


```python
numbers = [1,2,3,4,5]
result = list(map(lambda x : x +10,numbers))
print(result)
```

### Q9.Create a list names = ["rahul", "amit", "sneha"]. Use map() with a lambda function to convert all the names into uppercase and print the result as a list.

```python
names = ["rahul","amit","sneha"]
result = list(map(lambda name:name.upper() , names))
print(result)
```

### Q10.Create a list prices = [100, 200, 300, 400]. Use map() with a lambda function to apply a 10% discount on every price. Print the result as a list, then as a tuple.

```python
prices = [100,200,300,400]

result = map(lambda price :price -(price* 10/100),prices)
print(list(result))
result = map(lambda price :price -(price* 10/100),prices)
print(tuple(result))
```
