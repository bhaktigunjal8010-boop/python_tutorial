### 3.Create a class named Circle that takes radius as a parameter. Add a normal method area() that returns 3.14 × radius × radius. Create an object, and show that area() runs only when you call it, while __init__() runs automatically

```python
class Circle:
  def __init__(self,radius):
    print("object created")  ##init()runs automatically
    self.radius = radius
  def area(self):
    return 3.14 * self.radius *self.radius

obj=Circle(5)
obj.area() ## area runs when it called   
```

### 6.Create a class named Product with attributes name and price. Add a method apply_discount(percent) that reduces the price by the given percent and updates self.price. Print the price before and after calling the method

```python
class Product:
  def __init__(self,name,price):
    self.name = name
    self.price = price

  def apply_discount(self,percent):
    self.price = self.price - (self.price * percent / 100)

p1 = Product("laptop",50000)
print("before discount price :",p1.price) 

p1.apply_discount(20)
print("after discount price :",p1.price)
```

### 8.Create a class named Temperature that takes celsius as a parameter. Add a method to_fahrenheit() that returns (celsius × 9/5) + 32, and another method show() that calls self.to_fahrenheit() inside it and prints both values.



```python
class Temperature:
  def __init__(self,celcius):
    self.celcius = celcius

  def to_fahrenheit(self):
    return (self.celcius * 9/5) + 32

  def show(self):
    print("temperature in celcius",self.celcius)

    print("temperature in fahrenheit",self.to_fahrenheit())  

t1 =Temperature(20)
t1.show()      
```

### 9.Create a class named Calculator that takes two numbers as parameters. Add methods add(), subtract() and multiply(). Add one more method show_all() that calls all three methods using self and prints their results.



```python
class Calculator:
  def __init__(self,num1,num2):
    self.num1 =num1
    self.num2=num2

  def add(self):
    return self.num1 + self.num2

  def subtract(self):
    return self.num1 - self.num2

  def multiply(self):
    return self.num1 * self.num2

  def show_all(self):
    print("addition of numbers is : ",self.add())
    print("subraction of numbers is : ",self.subtract())
    print("multiplication of numbers is : ",self.multiply())
num = Calculator(20,5)
num.show_all()           
```
