### 4.Write a calculator program with four separate functions: add(), sub(), mul() and div(). Each function should take two numbers and return the result. Call all four functions with the numbers 20 and 5.

```python
def add(a,b):
   sum = a + b
   return sum
result = add(20,5)
print(result)

def sub(c,d):
   sum = c - d
   return sum
result = sub(20,5)
print(result)

def mul(e,f):
   sum = e * f
   return sum
result = mul(20,5)
print(result)

def div(g,h):
   sum = g / h
   return sum
result = div(20,5)
print(result)
```

### 8.Using the greet(name, age) function from Q7, call it using keyword arguments, once as greet(name="rahul", age=20) and once as greet(age=20, name="rahul").
