### Q2.Create a variable total_marks = 450 and a variable subjects = 0. Write a program that calculates the average marks using try and except. If ZeroDivisionError occurs, print "Number of subjects cannot be zero". Print "Result calculation completed" at the end.

```python
total_marks = 450
subjects=0
try:
  result = total_marks / subjects
  print("average marks :",result)
except ZeroDivisionError as err:
  print("number of subjects cannot be zero",err)

print("result calculation completed")    
```

### Q3.Write a function split_bill(amount, people) that returns the amount each person has to pay. Handle ZeroDivisionError inside the function. Call it once with (1200, 4) and once with (1200, 0).

```python
def split_bill(amount,people):
  try:
     return amount / people
     print("amount each person has to pay")
  except ZeroDivisionError as err:
     print("number of people cannot be zero",err)
     return 0

print(split_bill(1200,4))
print(split_bill(1200,0))  
```

### Q6.Calculate 10 * 400 and 10.0 * 400 separately using try and except. Which one works and which one raises an OverflowError? Write the reason in a comment.

```python
try:
  result1 = 10*400
  print(result1)
except OverflowError as err:
  print(err)  

try:
  result2 = 10.0*400
  print(result2)
except OverflowError as err:
  print(err)    

  # Both operations work because the values are small and do not exceed Python's numeric limits.Integers (10) can handle very large values, while floats (10.0) have a limited range.An OverflowError occurs only when a floating-point result becomes too large.
```

### Q10.Write a function safe_divide(a, b) that handles both ZeroDivisionError and TypeError using two separate except blocks. Call it with (50, 5), (50, 0) and (50, "5"), and print the result or error message for each call.

```python
def safe_divide(a,b):
  try:
    result = a/b
    print("result",result)
  except ZeroDivisionError as err:
    print(err)  
  except TypeError as err:
    print(err)
    

safe_divide(50,5)
safe_divide(50,0)
safe_divide(50,"5")

 
```
