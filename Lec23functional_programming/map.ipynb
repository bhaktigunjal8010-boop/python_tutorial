## 2.Map Function

#### Q1.Square of no.s using map function

```python
numbers = [1,2,3,4,5]
def square (x):
  return x ** 2
sq_nos = map(square,numbers)
print(sq_nos) 
```

```python
numbers = [1,2,3,4,5]
def square (x):
  return x ** 2
sq_nos = map(square,numbers)
print(list(sq_nos)) 
```

#### Q2.add 10 in every value and result new list using map

```python
numbers = [1,2,3,4,5]
def add(x):
  return x + 10
result = list(map(add , numbers))   
print(result)
```

#### Using lambda

```python
numbers = [1,2,3,4,5]
result = list(map(lambda x : x +10,numbers))
print(result)
```
