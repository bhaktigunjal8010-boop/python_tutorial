# Functional Programming

## 1.Lambda

#### Q.Create a single lambda function that adds two numbers

```python
add = lambda x ,y : x + y
print(add(5,10))
```

#### Q.check whether the number is even or not

```python
is_even = lambda x : x % 2 == 0
print(is_even(6))
```

```python
is_even = lambda x : x % 2 == 0
print(is_even(7))
```

#### Q3.square of number

```python
square = lambda x : x ** 2
print(square(4))
```

#### Q4.Concatenation of string using lambda

```python
concat = lambda str1 , str2 : str1 + str2
print(concat("hello ","world!"))
```
