## Arithmetic Operators

### 1.Addition(+)

```python
a = 10
b = 20
c = a + b
print(c)
```

### 2.Subtraction (-)

```python
a = 20
b= 10
c = a - b
print(c)
```

### 3.Multiplication(*)

```python
a= 10
b= 20
c= a * b
print(c)
```

### 4.Division(/)

```python
a = 10
b = 20
c= a/b
print(c)
```

### 5.floor division(//)

```python
a = 10
b = 20
c= a//b
print(c)
```

### 6.Module(%) [remainder]

```python
a = 10
b = 20
c = a % b
print(c)
```

```python
a = 10
b = 5
c = a % b
print(c)
```

### 7.Exponentiation(**) [raise to power]

```python
a = 10
b=2
c = a ** b
print(c)
```
