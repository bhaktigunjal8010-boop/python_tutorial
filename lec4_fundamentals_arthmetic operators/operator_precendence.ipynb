## Operator Precendence in Python

### 1.Parenthese ( )

```python
##with parenthese
a = (2 + 3) * 4
print(a)
```

```python
##without parenthese
a = 2 + 3 * 4
print(a)
```

### 2.Exponentiation (**)

```python
a = 2 + 3 ** 2
print(a)
```

### 3.Multiplication,division,floor division,module [ *, /,//,%]

```python
 a = 70 * 5 / 50 // 30 % 2
 print(a)
```

### 4.Addition, Subtraction (+,-)

```python
a = 10 - 2 + 3
print(a)
```
