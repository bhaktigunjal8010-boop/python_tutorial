## Homogenous Element

```python
tuple1 = (10,20,30,40,50)
```

```python
print(tuple1)
```

## Hetrogenous Element

```python
tuple2 = (10 , 10.2, "nagpur", True)
```

```python
print(tuple2)
```

## Duplicate Element

```python
tuple1 = (10,20,30,40,50,10)
```

```python
print(tuple1)
```

```python
tuple2 = ()
```

```python
print(tuple2)
```
