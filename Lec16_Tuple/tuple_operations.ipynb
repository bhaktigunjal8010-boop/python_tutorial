# Basic Tuple Operation

## 1.Immutability

```python
tuple=(10,20,30,40,50)
tuple[0] = 11
print(tuple)
```

## 2.Concatenation

```python
tuple1 =(10,20,30)
tuple2 = (40,50,60)
tuple3 = tuple1 + tuple2
print(tuple3)
```

## 3.Conversion

### 1]Converting string into tuple

```python
str1 = "hello"
```

```python
tuple1 = tuple(str1)
print(tuple1)
```

### 2]converting list into tuple

```python
list1=[10,20,30]
```

```python
tuple2=tuple(list1)
print(tuple2)
```

### 3]converting set into tuple

```python
set1= {10,20,30}
```

```python
tuple1 = tuple(set1)
print(tuple1)
```

### 4]converting dictionary into tuple

```python
dict1 = {"name":"ganesh","age":25}
```

```python
tuple1 = tuple(dict1)
print(tuple1)
```

### 5]Converting range into tuple

```python
range1 = range(1,11,1)
```

```python
tuple1 = tuple(range1)
print(range1)
```

## 4.Aliasing

```python
tuple1 = (10,20,30,40)
tuple2 = tuple1
print(tuple2)
```

```python
print(id(tuple1))
print(id(tuple2))
```

## 5.searching 'in' operator

```python
tuple1 = (10,10.1,2+3j,"ganesh",True)
```

```python
print(10 in tuple1)
```

```python
print("mahesh" in tuple1)
```

```python
print("ganesh"in tuple1)
```

## 6.Identity 'is' operator

```python
tuple1 = (10,20,30)
```

```python
tuple2 = tuple1
print(tuple1 is tuple2)
```

## 7.comparision

```python
tuple1 = (10,20,30)
tuple2 =(10,20,40,50)
```

```python
print(tuple1 > tuple2)
print(tuple1 < tuple2)
```

## 8.Emptiness 'not' operator

```python
tuple1 = ()
print(not tuple1)
```

```python
tuple2 =(10,20,30)
print(not tuple2)
```
