# Basic Tuple Operation

## Immutability

```python
tuple=(10,20,30,40,50)
tuple[0] = 11
print(tuple)
```

## Concatenation

```python
tuple1 =(10,20,30)
tuple2 = (40,50,60)
tuple3 = tuple1 + tuple2
print(tuple3)
```
