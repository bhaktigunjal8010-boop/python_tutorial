## class

```python
class vehical:
  def __init__(self,colour,top_speed,milage):  #constuctor
    self.colour = colour ,  #attribute1
    self.top_speed = top_speed , #attribute2
    self.milage = milage      #attribute3
  def display_info(self):  #method1
    print("colour is",self.colour)
    print("milage is",self.milage)
    print("top speed is",self.top_speed)  
```

```python
honda = vehical("grey",180,20)
honda.display_info()
```

```python
print("colour of car is", honda.colour)
```

```python
class vehical:
  def __init__(self):  #constuctor
    self.colour = "black"  #attribute1
    self.top_speed = 150 #attribute2
    self.milage = 15      #attribute3
  def display_info(self):  #method1
    print("colour is",self.colour)
    print("milage is",self.milage)
    print("top speed is",self.top_speed)  
```

```python
tata = vehical()  #object 
print(tata.top_speed)
print(tata.colour)
print(tata.milage)

tata.display_info()
```
