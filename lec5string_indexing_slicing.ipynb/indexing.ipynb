## Indexing

### 1.Positive Indexing

```python
str1 = "hello"
```

```python
print(str1[0])
print(str1[1])
print(str1[2])
print(str1[3])
print(str1[4])
```

## 2.Negative Indexing

```python
str1 = "Hello"
```

```python
print(str1[-5])
print(str1[-4])
print(str1[-3])
print(str1[-2])
print(str1[-1])
print(str1[-6])
```
