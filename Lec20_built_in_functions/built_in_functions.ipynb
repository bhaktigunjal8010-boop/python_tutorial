## 7.sorted( )

```python
dict7 = {40: "a", 20: "b", 30: "c", 10: "d"}
```

```python
print(sorted(dict7))
```

```python
for key in sorted(dict7):
    print(key, ":", dict7[key])
```

```python
## option2
list1 = sorted(dict7)
```

```python
for key in list1:
    print(key,":", dict7[key])
```

## 8.reversed()

```python
dict8 = {10: "a", 20: "b", 30: "c", 40: "d"}
```

```python
for key in reversed(dict8):
    print(key,":", dict8[key])
```

```python
## option2
list1 = list(reversed(dict8))
```

```python
for key in list1:
  print(key, ":", dict8[key])
```
