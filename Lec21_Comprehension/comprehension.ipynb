# Comprehension

## 1.List comprehension

### Q. create a list of square of no. from 1 to 5

#### 1]without comprehension

```python
for i in range(1,6,1):
  print(i**2)
```

#### 1.1] with comprehension

```python
list = [ i**2 for i in range(1,6,1)]
```

```python
print(list)
```

#### 2]convert words into uppercase

```python
animals = ["cat", "dog", "rabbit", "cat", "dog"]
```

```python
list = [i.upper() for i in animals]
```

```python
print(list)
```

```python
list = [i.upper() for i in animals if len(i)<=3]
```

```python
print(list)
```

#### 3]append( )   to add item

```python
new_animals = [i.upper()for i in animals if len(i)<=4]
```

```python
print(new_animals)
```

## 2.Set comprehension

### Q.create a set having squares of numbers from 1 to 5

## 1]with comprehension

```python
square_set = {i**2 for i in range(1,6,1)}
```

```python
print(square_set)
```

```python
square_set = {i**2 for i in range(1,6,1) if i<=3}
```

```python
print(square_set)
```

## 1.1]without comprehension

```python
set1 = set()
for i in range(1,6,1):
    if i<=3:
        set1.add(i**2)
        print(set1)
```

### 2]remove duplicate letters from string

```python
text = "asbgcbug"
```

```python
set1 = {i.upper() for i in text}
```

```python
print(set1)
```

```python
set2 = set(text)
```

```python
print(set2)
```

## 3.Dictionary Comprehension

### Q.create a dictionary having squares of no. from 1 to 5 where key is no. & value is square 

```python
dict1 ={i : i**2
        for i in range(1,6,1)}
```

```python
print(dict1)
```

```python
dict1 ={i : i**2
        for i in range(1,6,1)if i<=3}
```

```python
print(dict1)
```

### Q. create a dictionary where key is the animal name & value is the length of animal name

```python
animals = ["cat", "dog", "rabbit", "horse", "elephant"]
```

```python
dict = {i : len(i) for i in animals}
```

```python
print(dict)
```
