## 2.Set comprehension

### Q.create a set having squares of numbers from 1 to 5

## with comprehension

```python
square_set = {i**2 for i in range(1,6,1)}
```

```python
print(square_set)
```

```python
square_set = {i**2 for i in range(1,6,1) if i<=3}
```

```python
print(square_set)
```

## without comprehension

```python
set1 = set()
for i in range(1,6,1):
    if i<=3:
        set1.add(i**2)
        print(set1)
```
