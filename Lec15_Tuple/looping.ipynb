# Looping in Tuple

## While Loop

```python
tuple = (10,20,30,40,50)
index = 0
while index < len(tuple):
    print(tuple[index])
    index =index + 1
```

```python
tuple = (10,20,30,40,50)
index = 3
while index < len(tuple):
    print(tuple[index])
    index =index + 1
```

## For Loop

```python
tuple = (10,20,30,40,50)
for var in tuple:
    print(var)
```

## Enumerate() Function

```python
tuple = (10,20,30,40,50)
for index,value in enumerate(tuple):
    print(index,value)
```
