## Slicing

### Slicing Syntax :
                       tuple_name = [start : end+1 :step]

```python
tuple = (10,20,30,40,50)
```

```python
print(tuple)
```

```python
print(tuple[0 :1 :1])
```

```python
print(tuple[1 :2 :1])
```

```python
print(tuple[2 :3 :1])
```

```python
print(tuple[3 :4 :1])
```

```python
print(tuple[4 :5 :1])
```

```python
print(tuple[5:6 :1])
```

```python
print(tuple[ : :])
```

```python
print(tuple[::2])
```
