# Built in functions in Set

## 1.len( )

```python
set1 = {10,20,30,40,50}
```

```python
print(len(set1))
```

## 2.max( )

```python
set2 = {30,40,50,60,70}
```

```python
print(max(set2))
```

## 3.min( )

```python
set3 = {40,30,50,60,70}
```

```python
print(min(set3))
```

## 4.sorted( )

```python
set4 = {10,20,30,40,50}
print(set4)
```

```python
print(sorted(set4))
```

## 5.any( )

```python
set5 = {0,0,0,0,0,1}
```

```python
set5 = {0,0,0,0,0,0}
```
