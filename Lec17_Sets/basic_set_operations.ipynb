# Basic set operations

## 1.Mutability

### 1]add( )

```python
set1 = {10,20,30,40,50}
```

```python
set1.add(60)
print(set1)
```

### 2]remove

```python
set1.remove(30)
```

```python
print(set1)
```

## 2.Conversion

### 1]converting string into set

```python
str1 = "hello"
print(str1)
```

```python
set1 = set(str1)
```

```python
print(set1)
```

### 2]converting list into set

```python
list1 = [10,20,30,40,50]
print(list1)
```

```python
set2 = set(list1)
```

```python
print(set2)
```

### 3]converting tuple into set

```python
tuple1 = (10,20,30,40,50)
print(tuple1)
```

```python
set3 = set(tuple1)
```

```python
print(set3)
```

### 4]converting dictionary into set

```python
dict1 = {"name":"John", "age":30, "city":"New York"}
print(dict1)
```

```python
set4 = set(dict1)
```

```python
print(set4)
```

### 5]converting range into set

```python
range1 = range(1,11,1)
print(range1)
```

```python
set5 = set(range1)
```

```python
print(set5)
```

## 3.Aliasing

```python
set1 = {10,20,30,40,50}
print(set1)
```

```python
set2 = set1
print(set2)
```

```python
print(id(set1))
print(id(set2))
```

```python
set2.add(60)
print(set1)
print(set2)
```

## 4.Cloning

```python
set1 = {10,20,30,40,50}
print(set1)
```

```python
set2 = set1.copy()
print(set2)
```

```python
print(id(set1))
print(id(set2))
```

```python
set2.add(60)
print(set1)
print(set2)
```

## 5.Searching

```python
set1 = {10,20,30,40,50}
print(set1)
```

```python
print(20 in set1)
```

```python
print(60 in set1)
```

## 6.Identity

```python
set1 = {10,20,30,40,50}
print(set1)
```

```python
## using aliasing 
set2 = set1
print(set2)
```

```python
print(id(set1))
print(id(set2))
```

```python
print(set1 is set2)
```

```python
set3 = {30,40,50,60,70}
print(set3)
```

```python
##using cloning
set4 =set3.copy()
print(set4)
```

```python
print(id(set3))
print(id(set4))
```

```python
print(set3 is set4)
```
