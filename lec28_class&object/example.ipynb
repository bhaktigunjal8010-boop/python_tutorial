## Example1:

```python
class student:
  def __init__(self,name,age):
    print(id(self))
    self.name = name
    self.age = age
  def display_info(self):
    print("student name is : ",self.name)
    print("student age is : ", self.age)

s1=student("rahul",20)
print(s1.name)
print(s1.age)
print(id(s1))

s2 = student("akshay",25)
s2.display_info()
print(id(s2))
```

## Example2

```python
class school :
  def __init__(self,school_name,city):
    self.school_name = school_name
    self.city = city

  def display_info(self):
    print("school name is : ",self.school_name)
    print("school city is : ",self.city)

school1 = school("ABC School","pune")
school2 = school("XYZ School","mumbai")

print(school1.school_name)
print(school1.city)

print(school2.school_name)
print(school2.city)

school1.display_info()
school2.display_info()
```
