## Q1.What is functional programming in Python? Name the functions covered under it.

Functional programming in Python is a programming style where programs are built by using **functions** to perform operations on data. It focuses on using functions as objects, passing functions as arguments, and returning functions from other functions.

The main functions commonly covered under functional programming in Python are:

1. `map()` – applies a function to each item in an iterable.
2. `filter()` – selects items from an iterable based on a condition.
3. `reduce()` – combines all items of an iterable into a single value. It is available in the `functools` module.
4. `lambda` – creates a small anonymous function.

In short:

`map()` → transforms values
`filter()` → filters values
`reduce()` → reduces values to one result
`lambda` → creates a small anonymous function.


### Q2.What is a lambda function? Why is it called an anonymous function?

A **lambda function** is a small, one-line function in Python that is created using the `lambda` keyword.

It is called an **anonymous function** because it usually does not have a function name like a normal function defined using `def`.

Example:

```python
square = lambda x: x * x

print(square(5))
```

Output:

```text
25
```

Here, `lambda x: x * x` is a lambda function.

In short: A lambda function is a **small, one-line, anonymous function** used for simple operations.


### Q3.Explain the syntax of a lambda function

The syntax of a lambda function is:

```text
lambda arguments: expression
```

Example:

```python
square = lambda x: x * x
print(square(5))
```

Output:

```text
25
```

Here:

* `lambda` → keyword used to create a lambda function.
* `x` → argument or parameter.
* `:` → separates the arguments from the expression.
* `x * x` → expression that is evaluated and returned.

In short:

`lambda arguments: expression`

A lambda function can have **any number of arguments**, but it contains only **one expression**.


### Q4.What are the main characteristics of a lambda function?

The main characteristics of a lambda function are:

1. It is an **anonymous function**, meaning it usually does not have a name.
2. It is created using the **`lambda` keyword**.
3. It is generally written in **one line**.
4. It can accept **multiple arguments**.
5. It contains **only one expression**.
6. The result of the expression is **automatically returned**.
7. It is mainly used for **small and simple operations**.

Example:

```python
square = lambda x: x * x
print(square(5))
```

Output:

```text
25
```

In short: A lambda function is a **small, one-line, anonymous function that automatically returns the result of its single expression**.


### Q5.What is the difference between a normal function (def) and a lambda function?

The difference between a normal function (`def`) and a lambda function is:

Normal function (`def`):

* It is defined using the `def` keyword.
* It can contain multiple statements.
* It can have multiple expressions and a `return` statement.
* It usually has a name.
* It is suitable for larger and more complex tasks.

Lambda function:

* It is defined using the `lambda` keyword.
* It contains only one expression.
* The expression result is automatically returned.
* It is usually anonymous.
* It is mainly used for small and simple tasks.

Example of a normal function:

```python
def square(x):
    return x * x

print(square(5))
```

Example of a lambda function:

```python
square = lambda x: x * x

print(square(5))
```

Output for both:

```text
25
```

In short: `def` is used for **regular and complex functions**, while `lambda` is used for **small, simple, one-line functions**.


### Q6.Do we need to write the return keyword in a lambda function? Explain why.

No, we do not need to write the `return` keyword in a lambda function.

A lambda function **automatically returns the result of its expression**.

Example:

```python
square = lambda x: x * x
print(square(5))
```

Output:

```text
25
```

In short: **Lambda automatically returns the expression's result, so `return` is not required.**


### Q7.Can a lambda function take more than one argument? Give an example.

Yes, a lambda function can take more than one argument.

Example:

```python
add = lambda x, y: x + y
print(add(10, 20))
```

Output:

```text
30
```

Here, `x` and `y` are two arguments.


### Q8.What is the map() function? Explain its syntax.

The `map()` function applies a function to **each item of an iterable** and returns the results.

Syntax:

```text
map(function, iterable)
```

Example:

```python
numbers = [1, 2, 3]
result = map(lambda x: x * 2, numbers)

print(list(result))
```

Output:

```text
[2, 4, 6]
```

In short: `map()` → applies a function to each value.


### Q9.What does map() return? How do you convert the result into a list, set or tuple?

`map()` returns a **map object (iterator)** containing the results.

You can convert it into a list, set, or tuple using `list()`, `set()`, or `tuple()`.

Example:

```python
numbers = [1, 2, 3]
result = map(lambda x: x * 2, numbers)

print(list(result))
```

Output:

```text
[2, 4, 6]
```

In short:

`map()` → returns a map object
`list(result)` → list
`set(result)` → set
`tuple(result)` → tuple


### Q10.Can we use both a normal function and a lambda function with map()? Give an example of each.

Yes, we can use both a normal function and a lambda function with `map()`.

Normal function:

```python
def square(x):
    return x * x

numbers = [1, 2, 3]
print(list(map(square, numbers)))
```

Output:

```text
[1, 4, 9]
```

Lambda function:

```python
numbers = [1, 2, 3]
print(list(map(lambda x: x * x, numbers)))
```

Output:

```text
[1, 4, 9]
```

In short: `map()` can work with both **normal functions and lambda functions**.
