## Variable Length Arguments

### Q1.What are variable length positional arguments? When do we use them?

Variable-length positional arguments allow a function to accept any number of positional arguments.

In Python, we use *args to define variable-length positional arguments.

Example:

def add(*args):
    return sum(args)

print(add(10, 20))
print(add(10, 20, 30, 40))

Output:

30
100

When do we use them :
We use variable-length positional arguments when we do not know in advance how many arguments will be passed to a function.

Here, args is stored as a tuple.

### Q2.What is *args? In which data type does Python store the values received in *args?

*args is used to pass a variable number of positional arguments to a Python function.

Python stores all the values received through *args in a tuple.

Example:

def display(*args):
    print(args)

display(10, 20, 30)

Output:

(10, 20, 30)

### Q3.What is the difference between a normal positional argument and *args?

Normal positional arguments accept a fixed number of arguments, while *args accepts a variable number of positional arguments.

Normal positional arguments are stored in separate variables, whereas *args stores all received values in a tuple.

Example:
def add(a, b):
return a + b

def add(*args):
return sum(args)

### Q4.Is it compulsory to use the name "args" with *args? What is actually important, the name or the * symbol?

No, it is not compulsory to use the name args with *.

The * symbol is important. The name can be anything.

Example:

def add(*numbers):
print(numbers)

add(10, 20, 30)

Output:
(10, 20, 30)

Here, numbers works the same way as args.

### Q5.Can a function with *args be called with zero arguments? What will args contain in that case?

Yes, a function with *args can be called with zero arguments.

In that case, args will contain an empty tuple.

Example:

def greet(*args):
print(args)

greet()

Output:
()

### Q6.What are variable length keyword arguments? When do we use them?

Variable-length keyword arguments allow a function to accept any number of keyword arguments.

In Python, we use **kwargs to define variable-length keyword arguments.

Example:

def display(**kwargs):
    print(kwargs)

display(name="Bhakti", age=22, city="Pune")

Output:

{'name': 'Bhakti', 'age': 22, 'city': 'Pune'}

We use variable-length keyword arguments when we do not know in advance how many keyword arguments will be passed to a function.

The values received through **kwargs are stored in a dictionary.

### Q7.What is **kwargs? In which data type does Python store the values received in **kwargs?

`**kwargs` is used to pass a variable number of keyword arguments to a Python function.

Python stores the values received through `**kwargs` in a **dictionary**.

Example:

```text
def display(**kwargs):
    print(kwargs)

display(name="Bhakti", age=22)
```

Output:

```text
{'name': 'Bhakti', 'age': 22}
```

So, `**kwargs` stores the keyword arguments in a **dictionary**.


### Q8.What is the difference between *args and **kwargs?

*args is used to pass a variable number of positional arguments to a function. Python stores them in a tuple.

**kwargs is used to pass a variable number of keyword arguments to a function. Python stores them in a dictionary.

Example:

def test(*args):
print(args)

test(10, 20, 30)

Output:
(10, 20, 30)

We can also pass text (strings) using *args:

def test(*args):
print(args)

test("Hello", "Python", "Bhakti")

Output:
('Hello', 'Python', 'Bhakti')

def test(**kwargs):
print(kwargs)

test(name="Bhakti", age=22)

Output:
{'name': 'Bhakti', 'age': 22}

In short:

*args → variable positional arguments → stored in a tuple.

**kwargs → variable keyword arguments → stored in a dictionary.


### Q9.How can you loop through the values received in **kwargs?

You can loop through the values received in `**kwargs` using a `for` loop.

Example:

def display(**kwargs):
    for k, v in kwargs.items():
        print(k, ":", v)

display(name="Bhakti", age=22, city="Pune")
```

Output:


name : Bhakti
age : 22
city : Pune


Here, `kwargs.items()` returns the **key-value pairs** from the dictionary.

In short: We use a `for` loop with `kwargs.items()` to loop through the keys and values received in `**kwargs`.


### Q10.Can you use built-in functions like len() and sum() on *args? Give an example.

Yes, built-in functions like `len()` and `sum()` can be used with both `*args` and `**kwargs`.

For `*args`, Python stores the arguments in a tuple.

Example:

def calculate(*args):
print("Number of values:", len(args))
print("Sum of values:", sum(args))

calculate(10, 20, 30, 40)

Output:
Number of values: 4
Sum of values: 100

For `**kwargs`, Python stores the arguments in a dictionary.

Example:

def calculate(**kwargs):
print("Number of values:", len(kwargs))
print("Sum of values:", sum(kwargs.values()))

calculate(a=10, b=20, c=30)

Output:
Number of values: 3
Sum of values: 60

In short:

*args → stored in a tuple → `len(args)`, `sum(args)`

**kwargs → stored in a dictionary → `len(kwargs)`, `sum(kwargs.values())`
