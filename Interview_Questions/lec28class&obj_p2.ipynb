### 2. Why are append() and remove() called methods, while len() and sum() are called functions?

`append()` and `remove()` are called methods because they are called using a list object, whereas `len()` and `sum()` are called functions because they are built-in functions called directly by their names.

Example:

```
numbers = [10, 20, 30] 
numbers.append(40) 
 # Methodnumbers.remove(20) 
 # Methodprint(len(numbers)) 
 # Functionprint(sum(numbers)) # Function
```

Output:

```
3
80
```

In short: Methods are called using an object, while functions are called directly by their names.

### 3.What is a constructor in Python? Which method acts as the constructor?


In constructor we write configuration related code inside the constructor that we want to excecute automatically without user action. 

The `__init__()` method acts as the constructor in Python.

Example:

```
class Student:   
 def __init__(self, name):        
 self.name = name
 s1 = Student("Bhakti")
 print(s1.name)
```

Output:

```
Bhakti
```



### 4.What is the difference between the __init__() method and a normal method?

Difference between `__init__()` and a Normal Method:

| `__init__()` Method                                   | Normal Method                            |
| ----------------------------------------------------- | ---------------------------------------- |
| when we create object it automatically calls it       | explicitly call                          |
| syntax : `obj1 = class_name()`                        | syntax : `obj1.method_name()`            |
| user does not have control                            | user has control       |

Example:

```
class Student:   
 def __init__(self, name):        
 self.name = name    
 def display(self):        
 print(self.name)
 s1 = Student("Bhakti")
 s1.display()
```

Output:

```
Bhakti
```



### 5. Why do we write configuration-related code (like a database connection or API setup) inside the constructor?

We write configuration-related code inside the constructor (`__init__()`) because it runs automatically when an object is created. It initializes required resources, such as database connections or API settings, before using the object.

Example:

```
class Database:    
def __init__(self):        
self.host = "localhost"        
self.database = "student_db"
db = Database()
print(db.host)
print(db.database)
```

Output:

```
localhost
student_db
```

### 6. What is self? What does it refer to?

`self` is a reference to the current object of a class. It is used to access the object's attributes and methods.

Example:

```
class Student:   
 def __init__(self, name):       
  self.name = name   
   def display(self):       
    print(self.name)
    s1 = Student("Bhakti")
    s1.display()
```

Output:

```
Bhakti
```



### 7.How can you prove that self and the object refer to the same thing?

We can prove that `self` and the object refer to the same thing by using the `id()` function, which returns the identity of an object.

Example:

```
class Student:    
def show(self):       
 print(id(self))
 s1 = Student()
 print(id(s1))
 s1.show()
```

Output: Both `id(self)` and `id(s1)` print the same value.

.


### 9.How do you call one method of a class from another method of the same class?

We can call one method from another method of the same class using `self.method_name()`.

Example:

```
class Student:   
 def display(self):        
 print("Hello Student")   
  def show(self):        
  self.display()
  s1 = Student()
  s1.show()
```

Output:

```
Hello Student
```



### 10.What happens if you forget to write self as the first parameter of a method and then call it using an object?

If we forget to write `self` as the first parameter of an instance method, Python automatically passes the object as an argument when the method is called using an object. This causes a `TypeError` because the method receives an unexpected argument.

Example:

```
class Student:    
def display():       
 print("Hello Student")
 s1 = Student()
 s1.display()
```

Output:

```
TypeError: Student.display() takes 0 positional arguments but 1 was given
```
