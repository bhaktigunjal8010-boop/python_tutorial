### Q4.How do you define and call a function in Python? Explain the syntax.

A function in Python is defined using the def keyword followed by the function name and parentheses. Parameters can be written inside the parentheses. The function body is written with indentation.

Syntax:

def function_name(parameters):
    statements
    return value

To call a function, write the function name followed by parentheses and provide the required arguments.

Example:

def add(a, b):
    return a + b

result = add(10, 20)
print(result)

### 5.What is the naming convention for function names in Python?

Naming Convention for Function Names in Python

In Python, function names are generally written in lowercase letters, and when the name contains multiple words, they are separated using underscores (_). This style is called snake_case.

Examples:

calculate_sum()
print_message()
find_average()
Rules:
1.Use lowercase letters.
2.Use underscores to separate words.
3.A function name cannot start with a number.
4.Do not use Python keywords such as if, for, class, etc.
5.Choose a meaningful name that describes what the function does.

### 6.What is a docstring? Why is it used in a function?

A docstring (documentation string) is a string written inside a function to describe what the function does. It is written immediately after the function definition, usually using triple quotes (""" """).
A docstring is used to:

Explain the purpose of the function.
Describe its parameters and return value.
Make the code easier to understand and maintain.
Provide documentation for other programmers.
Example:
def add(a, b):
    """Returns the sum of two numbers."""
    return a + b



### 7.What happens if two functions are defined with the same name in a program?

If two functions have the same name in Python, the second function definition replaces the first one.

When the function is called, Python executes the second function.

Example:

def calculate(a, b):
    return a + b

def calculate(a, b):
    return a - b

result = calculate(10, 5)
print(result)

Output:

5

### 8.What is the difference between parameters and arguments?

Parameters are variables that are written inside the parentheses when defining a function. They are used to receive values.

Arguments are the actual values that are passed to a function when calling it.

Example:

def add(a, b):
    return a + b

add(10, 20)

Here, a and b are parameters, while 10 and 20 are arguments.

### 9.What is the use of the return keyword? What does a function return if there is no return statement?

Use of return Keyword

The return keyword is used to send a value back from a function to the place where the function was called.

Example:

def add(a, b):
    return a + b

result = add(10, 20)
print(result)

Output:

30

If a function has no return statement, Python automatically returns None.

Example:

def show_message():
    print("Hello")

result = show_message()
print(result)

Output:

Hello
None


return sends a value back from the function. If there is no return, the function returns None automatically.

### 10.What is the difference between positional arguments, keyword arguments and default arguments?

1. Positional Arguments:
Arguments are passed according to the position/order of the parameters.

def add(a, b):
    return a + b

add(10, 20)

Here, 10 is assigned to a and 20 to b.

2. Keyword Arguments:
Arguments are passed using the parameter name.

def add(a, b):
    return a + b

add(a=10, b=20)

Here, the values are assigned using the names a and b, so their order does not matter.

3. Default Arguments:
A parameter is given a default value in the function definition. If no value is provided, the default value is used.

def greet(name="User"):
    print("Hello", name)

greet()

Output:

Hello User

Positional argument → Value is passed according to position.
Keyword argument → Value is passed using the parameter name.
Default argument → A default value is used when no value is provided.
