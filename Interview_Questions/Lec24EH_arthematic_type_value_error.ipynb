# Interview Questions Session 24

## Exception Handling

### Q1.What are the two types of errors we get while working in Python?

The two main types of errors in Python are:

1. **Syntax Error** – Occurs when the rules or syntax of Python are not followed.
   Example: Missing colon `:` or bracket.

2. **Runtime Error (Exception)** – Occurs while the program is running.
   Example: Dividing a number by zero or using an undefined variable.


### Q2.What is the difference between a syntax error and an exception (runtime error)?

Difference between Syntax Error and Exception (Runtime Error):

1. Syntax Error:

* Occurs when the Python syntax is incorrect.
* The program cannot start running until the error is fixed.
* Example: Missing colon `:`.

2. Exception (Runtime Error):

* Occurs while the program is running.
* The syntax is correct, but an unexpected problem occurs.
* Example: Dividing a number by zero.

In short:

Syntax Error → Error in the structure/syntax of the code.
Runtime Error → Error that occurs during program execution.


### Q3.What happens to the rest of the program when an exception occurs and it is not handled?

When an exception occurs and it is **not handled**, the program **stops immediately at the point where the exception occurs**.

* The remaining statements are **not executed**.
* Python displays an **error message (traceback)** showing the cause of the exception.

Example:

```python
print("Start")
x = 10 / 0
print("End")
```

Output:

```text
Start
ZeroDivisionError: division by zero
```

Here, `"End"` is not printed because the program stops when the exception occurs.


### Q4.What is exception handling? Why do we need it?

Exception handling is a way of **handling errors that occur during program execution** so that the program does not stop suddenly.

We need exception handling to:

* Prevent the program from crashing.
* Handle errors properly.
* Display a meaningful error message.
* Allow the program to continue running when possible.

Python uses `try`, `except`, `else`, and `finally` for exception handling.


### Q5.Explain the syntax of try and except. What type of code should be written inside the try block?

Syntax of `try` and `except`:

```python
try:
    # code that may cause an exception
except:
    # code to handle the exception
```

The **`try` block** should contain code that may cause an error or exception.

Example:

```python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)
except:
    print("An error occurred.")
```

Here, the code inside the `try` block may cause an exception, such as entering `0` or a non-numeric value. The `except` block handles the error.


### Q6.What is the use of "as err" in an except block?

The use of **`as err`** in an `except` block is to store the **exception/error information in a variable** named `err`.

Example:

```python
try:
    x = 10 / 0
except Exception as err:
    print(err)
```

Output:

```text
division by zero
```

Here, `err` contains information about the error.

**In short:** `as err` helps us **access and display the error message**.


### Q7.When is a ZeroDivisionError raised?

A **`ZeroDivisionError`** is raised when we try to **divide a number by zero**.

Example:

```python
x = 10
y = 0

print(x / y)
```

Output:

```text
ZeroDivisionError: division by zero
```

**In short:** `ZeroDivisionError` occurs when a number is divided by `0`.


### Q8.When is an OverflowError raised? Why does it occur with float values and not with int values?

An **`OverflowError`** is raised when a calculation produces a value that is **too large to be represented by the given numeric type**.

**Example with float:**

```python
import math

print(math.exp(1000))
```

Output:

```text
OverflowError: math range error
```

**Why does it occur with float and not usually with int?**

* **Float:** Python floating-point numbers have a **fixed maximum range**. If the result is too large, an `OverflowError` can occur.
* **Int:** Python integers can grow to **very large sizes** because they have arbitrary precision. Therefore, normal integer calculations usually do not cause `OverflowError`.

**In short:**
`float` → has a fixed range → can overflow.
`int` → can grow very large → usually does not overflow.


### Q9.When is a TypeError raised? Give two examples.

A **`TypeError`** is raised when an operation or function is used with an **inappropriate or incompatible data type**.

**Example 1:**

```python
result = "10" + 5
```

Output:

```text
TypeError
```

Because a string and an integer cannot be added directly.

**Example 2:**

```python
result = len(10)
```

Output:

```text
TypeError
```

Because `len()` cannot be used on an integer.

**In short:** `TypeError` occurs when an operation is performed on an **incorrect data type**.


### Q10.If an exception is handled using try and except, will the statements written after the try-except block be executed? Explain.

Yes. If an exception is **handled properly** using `try` and `except`, the statements written **after the try-except block will be executed**.

Example:

```python
try:
    x = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")

print("Program continues")
```

Output:

```text
Cannot divide by zero
Program continues
```

Here, the exception is handled by `except`, so the program continues and executes the statement after the `try-except` block.

**In short:**
Handled exception → Program continues.
Unhandled exception → Program stops.
