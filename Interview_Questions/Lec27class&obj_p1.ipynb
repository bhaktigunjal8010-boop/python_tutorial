### 2.What does "everything in Python is an object" mean? Give an example using a built-in data type.

Everything in Python is an object means that every value has a type and supports certain operations and methods.

Example:


x = 10print(type(x))


Output:


<class 'int'>

Here, `10` is an object of the built-in `int` data type.

### 4.What is a class? Why is it called a blueprint?

1\. Class: A class is a user-defined structure that contains data members and methods.

2\. Blueprint: A class is called a blueprint because it defines the properties and behaviour of the objects created from it.

### 5.What is an object? Why is it called an instance of a class?

1\. Object: An object is an entity that has data and behaviour.

2\. Instance of a Class: An object is called an instance of a class because it is created using the class as a blueprint.

### 6.What are the two components of a class? Explain the difference between attributes and methods.



The two components of a class are:

1\. Attributes: Attributes are variables that store the data or properties of an object.

2\. Methods: Methods are functions defined inside a class that describe the behaviour of an object.

### 7.What is the _init_() method? When is it called?



1\. `__init__()` Method: It is a special method in Python used to initialize the attributes of an object.

2\. When is it called? It is automatically called when an object of a class is created.

### 8.What is self? Why do we write it as the first parameter in every method of a class



1\. Self: `self` is a reference to the current object of a class in Python.

2\. Use of self: It is used to access the attributes and methods of the current object.

3\. First Parameter: We write `self` as the first parameter of an instance method so that Python can pass the current object to the method.

4\. Example: In `def show(self):`, `self` refers to the object that calls the method.

### 10.How do you access an attribute and call a method of an object? Write the syntax.



1\. Accessing an attribute: We use the dot (`.`) operator to access an object's attribute.

Syntax: `object_name.attribute_name`

2\. Calling a method: We use the dot (`.`) operator followed by parentheses to call an object's method.

Syntax: `object_name.method_name()`
