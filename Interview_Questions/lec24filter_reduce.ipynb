## FILTER AND REDUCE

### Q1.What is the filter() function in Python? Explain its syntax.

### What is `filter()` function in Python?

The **`filter()` function** is used to select elements from an iterable (like a list or tuple) based on a **condition**. It returns only the elements for which the condition is `True`.

### Syntax:

```python
filter(function, iterable)
```

* **function** → A function that checks each element and returns `True` or `False`.
* **iterable** → A list, tuple, or other collection of elements.

### Example:

```python
numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))
```

**Output:**

```text
[2, 4, 6]
```

Here, `filter()` selects only the **even numbers** from the list.


### Q2.What should the function passed to filter() return? How does filter() decide which elements to keep?

The function passed to `filter()` should return a **Boolean value (`True` or `False`)**.

* If the function returns **`True`**, the element is **kept**.
* If it returns **`False`**, the element is **removed**.

**Example:**

```python
numbers = [1, 2, 3, 4, 5]

result = filter(lambda x: x > 3, numbers)

print(list(result))
```

**Output:**

```text
[4, 5]
```

Here, `filter()` keeps only the elements for which `x > 3` is `True`.


### Q3.What does filter() return? How do you convert the result into a list?

The `filter()` function returns a **filter object (iterator)** containing the elements that satisfy the given condition.

To convert the result into a **list**, use the `list()` function.

python
numbers = [1, 2, 3, 4, 5]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))


**Output:**


[2, 4]


**In short:** `filter()` → returns a filter object → `list()` converts it into a list.


### Q4.What is the difference between map() and filter()?

Difference between map() and filter():

1. map():

* Applies a function to every element.
* Used to transform or modify elements.
* Returns a map object.
* Example: Converting numbers into their squares.

2. filter():

* Applies a condition to every element.
* Used to select elements that satisfy a condition.
* Returns a filter object.
* Example: Selecting only even numbers.

In short:
map() → Transforms elements.
filter() → Selects elements based on a condition.


### Q5.Can we use both a normal function and a lambda function with filter()? Give an example of each.

Yes, we can use both a **normal function** and a **lambda function** with `filter()`.

1. Using a normal function:

```python
def is_even(x):
    return x % 2 == 0

numbers = [1, 2, 3, 4, 5, 6]

result = filter(is_even, numbers)

print(list(result))
```

Output:

```text
[2, 4, 6]
```

2. Using a lambda function:

```python
numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))
```

Output:

```text
[2, 4, 6]
```

In short: `filter()` can use both **normal functions and lambda functions**.


### Q6.What is the reduce() function? Explain its syntax.

The `reduce()` function is used to **apply a function repeatedly to the elements of an iterable** and reduce them to a **single value**.

`reduce()` is available in the `functools` module.

Syntax:

```python
from functools import reduce

reduce(function, iterable)
```

Example:

```python
from functools import reduce

numbers = [1, 2, 3, 4]

result = reduce(lambda x, y: x + y, numbers)

print(result)
```

Output:

```text
10
```

Here, `reduce()` adds all the numbers and returns a **single value (10)**.

Import statement:


from functools import reduce


### Q8.Explain step by step how reduce() works on the list [1, 2, 3, 4, 5] with an addition function.


Step-by-step working of `reduce()` on `[1, 2, 3, 4, 5]` with an addition function:

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

result = reduce(lambda x, y: x + y, numbers)

print(result)
```

Step-by-step:

1. First: `1 + 2 = 3`
2. Then: `3 + 3 = 6`
3. Then: `6 + 4 = 10`
4. Finally: `10 + 5 = 15`

Output:

```text
15
```

So, `reduce()` combines all elements one by one and produces a **single final value: 15**.


### Q9.How many arguments must the function passed to reduce() take? Why?

The function passed to `reduce()` must take **two arguments**.

* The first argument is the **accumulated result** from the previous operation.
* The second argument is the **next element** in the iterable.
* `reduce()` combines these two values repeatedly until only one final value remains.

Example:

```python
from functools import reduce

numbers = [1, 2, 3, 4]

result = reduce(lambda x, y: x + y, numbers)

print(result)
```

Here, `x` and `y` are the two arguments used by the function.


### Q10.What is the difference between the output of map(), filter() and reduce()?

Difference between the output of `map()`, `filter()` and `reduce()`:

1. `map()`:

* Returns a **map object**.
* Applies a function to every element.
* Usually produces the **same number of elements** as the input.

2. `filter()`:

* Returns a **filter object**.
* Selects only the elements that satisfy a condition.
* Usually produces **fewer elements** than the input.

3. `reduce()`:

* Returns a **single final value**.
* Combines all elements using a function.
* Reduces the input to **one value**.

In short:

`map()` → transforms elements
`filter()` → selects elements
`reduce()` → combines elements into one value
